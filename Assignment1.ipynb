```python
df['currency_rate'] = df.apply(lambda row: row['price'] / row['usdprice'] if row['usdprice'] != 0 else None, axis=1)
df[['commodity', 'price', 'usdprice', 'currency_rate']]
```

<p style="font-family: 'Arial'; font-style: italic;color: black;font-size:14px;">

<p style="font-family: 'Arial';font-style: italic;font-weight:bold; color:#36AF51;text-shadow: 5px 5px 10px rgba(0,0,0,0.15);font-size:19px;">- Question 3: Create a new column called unified_price that represents the price in USD per kilogram for commodities measured by weight and the price per liter for commodities measured by volume.</p>

```python
def unified_price(row):
    unit_info = row['unit'].split()

    if len(unit_info) < 2 and unit_info[0] in ['KG', 'L']:
        return row['usdprice'] / 1
    elif len(unit_info) == 2 and unit_info[-1] in ['KG', 'L']:
        amount = float(unit_info[0])
        return row['usdprice'] / amount
    else:
        return None

df['unified_price'] = df.apply(unified_price, axis=1)

df[['unit', 'usdprice', 'unified_price']]
```

<p style="font-family: 'Arial'; font-style: italic;color: black;font-size:14px;"> 
The code provides the <strong>unified_price</strong> function that determines the common price for a given commodity measured in a certain unit. It divides the unit string and determines whether it has <strong>singular or divisional</strong> units: if it is just a single unit without a number (for example “KG” or “L”), it returns the usdprice; while in the case of with number (for example with “5 KG” it takes the usdprice number and divides it with that number). It does return None, if the format is not understood, while the function is used on the DataFrame to return a new column <strong>‘unified_price’</strong> with the calculated values.
</p>

<p style="font-family: 'Arial';font-style: italic;font-weight:bold; color:#36AF51;text-shadow: 5px 5px 10px rgba(0,0,0,0.15);font-size:19px;">- Question 4: Using the unified_price, find the maximum price reached during this period and for which commodity for both:
commodities measured by weight
commodities measured by volume.</p>

```python
weight = df[df['unit'].apply(lambda x: x.split()[-1] == 'KG')]

max_weight = weight['unified_price'].max()

max_weight_commodity = weight.loc[weight['unified_price'] == max_weight, 'unit'].values

volume = df[df['unit'].apply(lambda x: x.split()[-1] == 'L')]

max_volume = volume['unified_price'].max()

max_volume_commodity = volume.loc[volume['unified_price'] == max_volume, 'unit'].values

print(f"Max price for weight commodities: {max_weight} USD for {max_weight_commodity}")
print(f"Max price for volume commodities: {max_volume} USD for {max_volume_commodity}")
```

<p style="font-family: 'Arial'; font-style: italic;color: black;font-size:14px;"> 
The code selects from the Data Frame only those commodities which are examined in weight (KG) and volume (L) and assigns it to variables <strong>weight</strong> and  <strong>volume</strong>. It uses a <strong>max()</strong> function to determine the maximum available value of the unified_price for both classes of commodity and fetches the required commodity names corresponding to each price. The output for weight and volume category of the maximum price is also displayed in a single string where values of maximum prices are shown along with names of the products.
